# CheXpert Results Visualization

This notebook plots the results from Table 1 (CheXpert row) showing the performance of different calibration methods.

```python
import json
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Load the results
with open('../experiments/vision/results/json/aggregated_results_chexpert.json', 'r') as f:
    results = json.load(f)
```

```python
# Define method names and labels
methods = {
    'baseline': 'Original',
    'replace_mean': 'Replace',
    'patchcutout': 'PatchCutout',
    'temperature': 'Temperature',
    'platt': 'Platt',
    'arch_mod': 'Arch Mod',
    'mcal_ce': 'MCal CE'
}

# Extract data for each method (only include methods that exist in results)
data = {}
for key, label in methods.items():
    if key in results:
        if key == 'baseline':
            data[label] = results[key]['fraction_wise_results']['mean_argmax']
        else:
            data[label] = results[key]['fraction_wise_results_transformed']['mean_argmax']

# Create x-axis (fraction values)
if data:
    n_points = len(next(iter(data.values())))
    fractions = np.linspace(0, 1, n_points)
else:
    print("Warning: No data found in results")
```

```python
# Create the plot with log scale
if data:
    plt.figure(figsize=(6, 4.8))

    # Plot each method
    for label, values in data.items():
        plt.plot(fractions, values, marker='o', label=label, linewidth=2, markersize=4)

    plt.xlabel('Fraction of Features Removed', fontsize=14)
    plt.ylabel('KL Divergence (Mean Argmax)', fontsize=14)
    plt.title('CheXpert: Calibration Method Comparison', fontsize=16)
    plt.yscale('log')
    plt.legend(loc='best', fontsize=11)
    plt.grid(True, alpha=0.3)
    plt.tick_params(labelsize=12)
    plt.tight_layout()
    plt.show()
else:
    print("No data to plot")
```

```python
# Display summary statistics
if data:
    print("Summary Statistics (Mean KL Divergence across all fractions):")
    print("="*60)
    for label, values in data.items():
        mean_kl = np.mean(values)
        print(f"{label:15s}: {mean_kl:.6f}")
else:
    print("No data available for summary statistics")
```
